# Introduction

Emerging infectious diseases of zoonotic origin continue to threaten global health security, with over 60% of emerging human pathogens originating in animals. The COVID-19 pandemic, caused by a zoonotic betacoronavirus, emphasized the importance of understanding viral reservoirs in wildlife. Bats have been extensively studied as natural hosts of many zoonotic viruses, including several coronaviruses, but the role of rodents as viral reservoirs remains comparatively underexplored. This is despite their ecological ubiquity, vast species richness, and close association with human environments.

Rodents (order Rodentia) comprise over 2,200 species and represent the most speciose mammalian order. Their wide geographic distribution, adaptability to urban environments, and potential for high-density populations make them prime candidates for facilitating spillover events. Several major viral pathogens with significant human health impacts have emerged from rodent hosts, including hantaviruses, arenaviruses, and most recently, mpox virus, which has been linked to African rodent species.

Hantaviruses, carried by murid rodents, have caused multiple outbreaks of hantavirus pulmonary syndrome (HPS) in the Americas and hemorrhagic fever with renal syndrome (HFRS) in Eurasia. In particular, the Andes virus, endemic to South America, is of high concern due to its rare but documented capacity for person-to-person transmission. This distinguishes it from most other hantaviruses and raises significant concerns about its pandemic potential. Andes virus is primarily carried by *Oligoryzomys longicaudatus*, a rodent widely distributed across Argentina and Chile. Arenaviruses, another group of rodent-borne viruses, include several species responsible for severe hemorrhagic fevers. The Junin virus, endemic to Argentina, causes Argentine hemorrhagic fever (AHF) and is carried by *Calomys musculinus*. Historically associated with agricultural outbreaks, Junin virus infection has a high case-fatality rate if untreated, though a live attenuated vaccine (Candid #1) has been used with success. Its presence in a commonly encountered rodent in rural areas highlights the persistent risk posed by zoonotic viruses endemic to agricultural interfaces.

The recent reemergence of mpox (formerly monkeypox), a poxvirus that is believed to have originated in African rodents, has also renewed interest in rodent reservoirs. Though initially considered a rare zoonosis with limited human-to-human spread, mpox has evolved in ways that allow more sustained transmission, and wild rodents remain a suspected reservoir facilitating its maintenance in endemic regions.

Despite this pattern of rodent-associated viral emergence, coronaviruses found in rodents have received relatively little attention until recently. Within the genus *Betacoronavirus*, the subgenus *Embecovirus* contains several human coronaviruses (e.g., OC43, HKU1) that likely originated from rodents or ungulates. Rodent-associated *Embecoviruses* have been detected globally, but their diversity, evolutionary history, and zoonotic potential are still poorly understood.

Here, we report the discovery and phylogenetic characterization of two novel *Embecovirus*-like coronaviruses identified in *Peromyscus leucopus*, a widespread North American rodent species known for its ecological overlap with humans and its role as a reservoir for multiple tick-borne pathogens. Using metagenomic sequencing and genome-resolved phylogenetics, we demonstrate that these viruses form a deeply divergent lineage within the *Embecovirus* subgenus. Our results expand the known diversity of rodent-borne coronaviruses and support the growing consensus that rodents play a critical role in coronavirus evolution and emergence.

This work underscores the importance of sustained viral surveillance in rodent populations,  including those found in dense urban environments where humans and wildlife coexist in close proximity. As anthropogenic pressure continues to alter ecosystems and increase contact between humans and wildlife, identifying and characterizing the viral diversity harbored by rodents will be essential to anticipate and prevent future zoonotic spillover events.

## 2. Pancoronavirus screening and sequencing
Total RNA was obtained from tissues, feces, and swabs using the MagMax mirVana Total RNA isolation kit (Thermo Fisher Scientific) using the KingFisher Apex Purification System (Thermo Fisher Scientific). Prior to RNA extraction between 20 to 40 mg of sample was disrupted using the Bead Ruptor Elite (OMNI International) and 800 ul of lysis buffer. For the disruption, MagMAX Microbiome Bead Tubes were used for feces or rectal swabs and 500mg of ~2.5mm Zirconia/Silica beads (Biospec Products) were used for tissues and saliva and nasal swabs. The obtained RNA was reverse transcribed using the SuperScript VILO Master Mix (Thermo Fisher Scientific). CoV RNA was detected using a pancoronavirus nested PCR with primers targeting the conserved RdRp gene of CoVs as previously published (Watanabe et al., 2010) and using the Platinum II Taq Hot-Start DNA Polymerase (Thermo Fisher Scientific). After PCR, samples were run in a 1% agarose gel stained with GelRed (Biotium) and visualized in a Gel Doc XR System (Biorad).

Pan-CoV positive samples were sequenced via nanopore (PromethION, Oxford Nanopore) using the [ARTIC](https://www.protocols.io/view/artic-sars-cov-2-sequencing-protocol-v4-lsk114-bp2l6n26rgqe/v4) amplification protocol. Amplicons were barcoded (SQK-RBK-114.96) and sequenced overnight.

To investigate the evolutionary relationships of the newly identified rodent coronaviruses, we performed a phylogenetic analysis based on the RNA-dependent RNA polymerase (RdRp) region encoded by the NSP12 gene. Viral genomes for the two positive samples (SM009RE and SM004RE) were aligned against related betacoronaviruses (Table 1) and annotated using the NSP12 coordinates from Mouse Hepatitis Virus A59 (NC_048217). Nucleotide sequences corresponding to the NSP12 region were extracted from each genome in Geneious (Dotmatics) and aligned using Clustal Omega. 

Whole-genome nucleotide sequences of 24 coronaviruses were compiled and aligned to reconstruct their evolutionary relationships. Each genome was renamed using its GenBank accession number to ensure consistent downstream handling. The sequences were aligned using MAFFT v7.505 with the --auto option for automatic strategy selection and default gap penalties. The resulting multiple sequence alignment contained 34,596 positions. Phylogenetic inference was performed using IQ-TREE v2.2.6 under the GTR+G substitution model, which was selected automatically based on model testing. Ultrafast bootstrap approximation with 1000 replicates (-B 1000) was conducted to assess node support. The resulting tree was rooted using the outgroup method and exported in Newick format as a .treefile.
